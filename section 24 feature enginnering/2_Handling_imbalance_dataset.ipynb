## Handling Imbalanced Dataset

1. Up Sampling
2. Down Sampling

### 1. Up Sampling


```python
import pandas as pd 
import numpy as np

# set the random seed for reproductivity 
np.random.seed(123)

# create a dataframe with 2 classes 
n_samples=1000
n_class_0_ratio=0.9

n_class_0=int(n_samples * n_class_0_ratio)
n_class_1=n_samples-n_class_0
```

```python
n_class_0,n_class_1
```

```python
## CREATE MY DATAFRAME WITH IMBALANCED DATASET

class_0=pd.DataFrame({
    'feacture_1':np.random.normal(loc=0, scale=1, size=n_class_0),
    'feacture_2':np.random.normal(loc=1,scale=1, size=n_class_0),
    'target':[0]*n_class_0
})

class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature_2': np.random.normal(loc=2, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})
```

```python
class_1,class_0
```

```python
df=pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df.head(904)
# df.tail()
```

```python
df['target'].value_counts()
```

```python
## upsampling
df_minority=df[df['target']==1]
df_majority=df[df['target']==0]
```

```python
from sklearn.utils import resample
df_minority_upsampled=resample(df_minority,replace=True, #Sample With replacement
         n_samples=len(df_majority),
         random_state=42
        )
```

```python
df_minority.shape   #(100,5)
df_minority_upsampled.shape   #(900,5)
```

```python
df_minority_upsampled.head()
```

```python
df_upsampled=pd.concat([df_majority,df_minority_upsampled])
```
