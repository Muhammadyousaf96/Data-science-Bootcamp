What is Polymorphism?

Polymorphism means:

"Many forms"

In OOP:

One method name can be used in different ways.

Same function name → different behavior.

```python
# Base class 
class Animale:
    def speak(self):
        return f"Sound of the Animale"

# Derived class 1
class Dog(Animale):
    def speak(self):
        return "woof"    

#Derived class 2
class cat(Animale):
    def speak(self):
        return "meow"
 
#function that demonstrate the polymorphism
def animal_speak(Animale):  
    print(Animale.speak())
    
dog=Dog()     
dog.speak()
cat=cat()
# cat.speak()

```

Polymorphism with functions and methods:
===
Polymorphism with functions and methods means that the same function or method name can be used to perform different tasks based on the context in which it is used. This allows for greater flexibility and code reusability.For example, consider a function called "area" that calculates the area of different shapes. 

```python
# Base classs 
class Shape:
    def area(self):
        return "Area of the figure"

# Deriverd class 1
class rectangle(Shape):
    def __init__(self,width,height):
        self.width=width
        self.height=height
        
    def area(self):
        return self.width * self.height        
    
#Derived class 2
class circle(Shape):
    def __init__(self,radius):
        self.radius=radius
    
    def area(self):
        return 3.14*self.radius*self.radius
    
#function that demonstarion polymorphism 
def print_area(Shape):
    print(f"the are is {Shape.area()}")  
    
rectangle=rectangle(2,5)
rectangle.area()    
circle=circle(4)
circle.area()
```

Polymorphism with abstract classs
===
Polymorphism with abstract classes means that you can define a common interface for a group of related classes, and each class can provide its own implementation of that interface. This allows you to write code that can work with objects of different classes without needing to know their specific types. For example, you could define an abstract class called "Shape" with an abstract method called "area", and then create subclasses like "Circle", "Square", and "Triangle" that implement the "area" method in their own way. This way, you can use polymorphism to calculate the area of different shapes without needing to know the specific type of shape you are working with.

```python
from abc import ABC,abstractmethod

#define an abstract class
class Vechile(ABC):
    @abstractmethod
    def start_engin(self):
        pass
    
#Derived class 1
class Car(Vechile):
    def start_engin(self):
        print(f"The engine of car are start") 
        
#Derived method 2
class Motercyle(Vechile):
    def start_engin(self):
        print(f"the Motercycle are start ")        

#Creaate object of car and moterycle
car=Car()
motercycle=Motercyle()

car.start_engin() 
motercycle.start_engin()
```
