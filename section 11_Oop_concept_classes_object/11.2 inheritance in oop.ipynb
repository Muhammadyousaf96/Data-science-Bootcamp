What is Inheritance?
===
Inheritance means:
A class can inherit (reuse) properties and methods from another class.

In simple words:<br>
👨 Father → gives property to Son<br>
📦 Parent Class → gives features to Child Class<br>
This helps us reuse code and avoid repetition.<br>



```python
#base class parent class
class Animal:
    def __init__(self,name,age):
        self.name=name
        self.age=age
        
    def speak(self):
        print(f"The animal:{self.name} are speak")  
  
#Derived class or child class
class Dog(Animal):
    def __init__(self,name,age):
        super().__init__(name,age)
    def bark(self):
        print(f"the {self.name } are barking")
dog=Dog("hasky","how")        
dog.speak()
dog.bark()                  
```

```python
# Inherentance 
#Parent Class 
class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype
    
    def drive(self):
        print(f"You can Drive the car of enginetype: {self.enginetype}")
            
# car=Car(4,3,"petrol")            
# car.drive()        

# Child class 
class Tesla(Car):
    def __init__(self, windows, doors, enginetype,is_self_drive):
        super().__init__(windows,doors,enginetype)
        self.is_self_drive=is_self_drive
        
    def self_drive(self):
        print(f"the Car is self_drive:{self.is_self_drive}")   
        
tesla=Tesla(5,3,"electric",True)
# print(tesla)
# tesla.doors
tesla.self_drive()
tesla.enginetype
tesla.is_self_drive
```

Multiple Inheritance
===

```python
# when a class inhert from more than one base class 
#in multiple inhrtsnce use direct class name instant of super() method

# base class 1
class Animal:
    def __init__(self,name):
        self.name=name
    
    def Sound(self):
        print(f"The {self.name} are Sound as ...")
#Base class 2
class Pet:
    def __init__(self,owner):
        self.owner=owner
    
#Derived class
class Dog(Animal,Pet):
    def __init__(self,name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)
  
    def speak(self):
        return f"the {self.name} is say woof"
  
dog=Dog("hasky","krish")                            
dog.speak()
```
