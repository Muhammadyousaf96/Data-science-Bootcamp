Instance Variable
===

```python
# instance variable , the variable which belong to the object
# self  means this object and use the variable every where within the class in every methos or function in a class
# for instance variable we use a __init__ constructor and define the instance variable within the constructor with the self keyword
class Dog:
    def __init__(self,name,age):
        self.name=name
        self.age=age
dog1=Dog("max",5)
print(dog1)     
print(dog1.name)   
print(dog1.age)
dog2=Dog("lucy",6)
print(dog2.age)
print(f"Dog1:{dog1.name} and Dog2:{dog2.name}")
```

Instance methods
 ===

```python
# Instance method are thos which belng to the object 
class University:
    def __init__(self,name,location,Ranking):
        self.name=name
        self.location=location
        self.Ranking=Ranking
    # here we we define instance method 
    def get_info(self):
        print(f"Uni name :{self.name} ")
        print(f"uni location: {self.location}")
        print(f"uni Rank: {self.Ranking}")
        
        
#creating object of class university 
Uni1=University("nfc-iet-multan","multan","top-3")
print(Uni1.get_info())
        
```

Modling Bank Account
===

```python
# Modling Bank Acount where we use method like depost and withdraw ets
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.balance=balance
    
    def deposit(self,amount):
        self.balance+=amount
        print(f"The Amount {amount} are depoist to Your Acc & total balance:{self.balance}")
    
    def withdraw(self,amount):
        if amount>self.balance:
            print("insuffesnt fund")
        else:
            self.balance=self.balance-amount
            print(f"The Amount:{amount} are withdeaw from your acc & total balance:{self.balance}")
    def get_balance(self):
        return self.balance        
                      
#Creating objects 
account=BankAccount("M Yousaf",6000)   
print(account.owner)
print(account.balance)
account.deposit(2000)
account.withdraw(300)
account.get_balance()
```
