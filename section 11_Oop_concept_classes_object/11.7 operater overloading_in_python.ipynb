Operator Overloading in Python 🐍
===
Operator overloading lets you define how Python's built-in operators (+, -, *, ==, etc.) behave with your custom classes. 
This makes your 
objects feel natural and intuitive to use

Why Operator Overloading?
===
Without it, doing obj1 + obj2 would throw a TypeError. With it, you define exactly what that means for your data.

Example — A Vector Class (great for Data Science!)
===

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # + operator
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    # - operator
    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    # * operator (scalar multiplication)
    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)

    # == operator
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    # print-friendly output
    def __str__(self):
        return f"Vector({self.x}, {self.y})"

    # len() → magnitude (rounded)
    def __len__(self):
        return int((self.x**2 + self.y**2) ** 0.5)


# ---- Usage ----
v1 = Vector(2, 3)
v2 = Vector(1, 4)

print(v1 + v2)   # Vector(3, 7)
print(v1 - v2)   # Vector(1, -1)
print(v1 * 3)    # Vector(6, 9)
print(v1 == v2)  # False
print(len(v1))   # 3
```

A Data Science Example — Custom Dataset Class
===

```python
class Dataset:
    def __init__(self, data: list):
        self.data = data

    # Combine two datasets with +
    def __add__(self, other):
        return Dataset(self.data + other.data)

    # Number of records with len()
    def __len__(self):
        return len(self.data)

    # Check equality
    def __eq__(self, other):
        return self.data == other.data

    def __str__(self):
        return f"Dataset({self.data})"


d1 = Dataset([1, 2, 3])
d2 = Dataset([4, 5, 6])

combined = d1 + d2
print(combined)    # Dataset([1, 2, 3, 4, 5, 6])
print(len(d1))     # 3
print(d1 == d2)    # False
```

```python
# basic example 
class number:
    def __init__(self,value):
        self.value=value
    
    def __add__(self,other):
        return number(self.value + other.value)
    
    def __str__(self):    
        return str(self.value)
    
n1=number(32)    
n2=number(21)
print(n1+n2)
print(n1)
```

🧪 Multiple Operator Example
===

```python
class vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y
        
    def __add__(self, other):
        return vector(self.x + other.x, self.y + other.y)
    
    def __str__(self):
        return f"{self.x}, {self.y}"
    
        
v1=vector(3,4)
v2=vector(5,6) 

print(v1+v2)


# print(v1)     
# print(v2)  
# print(v1+v2)
        
```
