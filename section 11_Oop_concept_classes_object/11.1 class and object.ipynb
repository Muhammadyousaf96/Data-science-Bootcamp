```python
# ==============================
# CLASS & OBJECT EXAMPLE
# ==============================

class Student:
    
    # Constructor (Initialize object)
    def __init__(self, name, age):
        # Instance Variables
        self.name = name
        self.age = age
    
    # Instance Method
    def greet(self):
        print(f"Hello, my name is {self.name} and I am {self.age} years old.")
        

# Creating Objects
student1 = Student("Ali", 22)
student2 = Student("Sara", 25)

# Calling Instance Method
student1.greet()
student2.greet()
```

2️⃣ Clear Example of Instance Variables & Methods
===

```python
# ==============================
# INSTANCE VARIABLE & METHOD DEMO
# ==============================

class Car:
    
    def __init__(self, brand, model):
        self.brand = brand      # Instance Variable
        self.model = model      # Instance Variable
    
    def show_info(self):        # Instance Method
        print(f"Car Brand: {self.brand}")
        print(f"Car Model: {self.model}")


car1 = Car("Toyota", "Corolla")
car2 = Car("Honda", "Civic")

car1.show_info()
print()
car2.show_info()
```

✅ 3️⃣ Real OOP Model — Bank Account System 🏦
===

```python
# ==============================
# BANK ACCOUNT OOP MODEL
# ==============================

class BankAccount:
    
    def __init__(self, owner, balance=0):
        # Instance Variables
        self.owner = owner
        self.balance = balance
    
    # Instance Method - Deposit
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"{amount} deposited successfully.")
        else:
            print("Deposit amount must be positive.")
    
    # Instance Method - Withdraw
    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance!")
        elif amount <= 0:
            print("Withdrawal amount must be positive.")
        else:
            self.balance -= amount
            print(f"{amount} withdrawn successfully.")
    
    # Instance Method - Check Balance
    def check_balance(self):
        print(f"Account Holder: {self.owner}")
        print(f"Current Balance: {self.balance}")
        print("-" * 30)


# Creating Bank Account Object
account1 = BankAccount("Muhammad Yousaf", 1000)

account1.check_balance()

account1.deposit(500)
account1.check_balance()

account1.withdraw(300)
account1.check_balance()

account1.withdraw(2000)  # Trying to withdraw more than balance
```
