Abstraction in OOP (Python)
===
What is Abstraction?
Abstraction means hiding complex internal details and only showing the essential features to the user.

 You don't need to know how something works — just what it does.
 
Think of driving a car.
===
You use the steering wheel, accelerator, and brakes — you don't need to know how the engine, fuel injection, or transmission works internally.

Abstraction = showing only what is necessary, hiding the complex details.

In Python OOP:
— You define what a class should do (abstract methods)

— You hide how it does it (implementation in subclass)

Creating an abstract class ABC
===


```python
from abc import ABC, abstractmethod

# Abstract class — like a blueprint/template
class Animal(ABC):   # Inherits from ABC

    def breathe(self):           # Normal method ✅
        print("Inhale... Exhale...")

    @abstractmethod
    def sound(self):              # Abstract method ❗
        pass                       # No body here!

class cow(Animal):
    def sound(self):
        print("the cow are sounded")
    
co=cow()  
co.sound()  
# ❌ Cannot create object of abstract class
# a = Animal()  # TypeError: Can't instantiate abstract class!
```

Abstract method
===

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

    @abstractmethod
    def move(self):
        pass

# ✅ Child class MUST implement all abstract methods
class Dog(Animal):
    def sound(self):
        print("Woof! Woof!")

    def move(self):
        print("Dog runs on 4 legs")

class Bird(Animal):
    def sound(self):
        print("Tweet! Tweet!")

    def move(self):
        print("Bird flies with wings")

# ✅ Now we CAN create objects
d = Dog()
d.sound()  # Woof! Woof!
d.move()   # Dog runs on 4 legs

b = Bird()
b.sound()  # Tweet! Tweet!
```

Real example 
===

```python
from abc import ABC, abstractmethod

# Blueprint — every payment method MUST have these
class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

    @abstractmethod
    def refund(self, amount):
        pass

# CreditCard — implements its own way
class CreditCard(Payment):
    def pay(self, amount):
        print(f"Paid {amount} via Credit Card")

    def refund(self, amount):
        print(f"Refunded {amount} to Credit Card")

# PayPal — implements its own way
class PayPal(Payment):
    def pay(self, amount):
        print(f"Paid {amount} via PayPal")

    def refund(self, amount):
        print(f"Refunded {amount} to PayPal account")

# Usage — same interface, different behaviour!
for method in [CreditCard(), PayPal()]:
    method.pay(500)

# Output:
# Paid 500 via Credit Card
# Paid 500 via PayPal
```

Abstraction vs Encapsulation
===
Abstraction (what)

Hides complexity — shows only what the object does

Like a TV remote — you press a button, you don't know the internal circuits

Implemented via: ABC, @abstractmethod

Encapsulation (how)

Hides data — protects internal variables from outside access

Like a capsule pill — medicine is hidden inside, protected from air

Implemented via: _protected, __private
