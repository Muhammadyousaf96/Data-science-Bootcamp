Magic Methods in Python
===
What are Magic Methods?
===
Magic Methods are special built-in methods in Python that start and end with double underscores __like_this__. 

 They are also called Dunder Methods (Double UNDERscore).
Python calls them automatically — you don't call them yourself!

When you write a + b, Python secretly calls a.__add__(b)

When you write print(obj), Python secretly calls obj.__str__()

When you write len(obj), Python secretly calls obj.__len__()

Magic methods let you define how YOUR objects behave with built-in Python operations.

```python

#without the magic method 

class Book:
    def __init__(self,title):
        self.title=title
        
book=Book("python")
print(book)        
```

```python
#with magic method __str__()
class Book:
    def __init__(self,title):
        self.title=title
    
    def __str__(self):
        return f"Book:{self.title}"    
b=Book("java")
print(b)    
```

__init__ — the constructor (runs when object is created)
===

```python
class Student:
    def __init__(self, name, age, grade):
        # Runs automatically when Student() is called
        self.name  = name
        self.age   = age
        self.grade = grade
        print(f"Student {name} created!")

    def __del__(self):
        # Runs automatically when object is destroyed
        print(f"Student {self.name} removed!")

s1 = Student("Ali", 20, "A")  # → Student Ali created!
s2 = Student("Sara", 22, "B") # → Student Sara created!
# print(s1.name)
print(s2.name)
# del s1
print(s1.age)
```

__str__ vs __repr__ — how your object looks when printed
===

```python
class Phone:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

    def __str__(self):
        # For users — readable, clean output
        return f"{self.brand} phone — Rs.{self.price}"

    def __repr__(self):
        # For developers — detailed, unambiguous output
        return f"Phone(brand='{self.brand}', price={self.price})"

p = Phone("Samsung", 50000)

print(p)      # calls __str__  → Samsung phone — Rs.50000
print(repr(p)) # calls __repr__ → Phone(brand='Samsung', price=50000)
```

Arithmetic magic methods — make + - * / work on your objects
===

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

    # def __add__(self, other):
    #     # Called when we write v1 + v2
    #     return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        # Called when we write v1 - v2
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self, num):
        # Called when we write v1 * 3
        return Vector(self.x * num, self.y * num)

    def __len__(self):
        # Called when we write len(v1)
        return int((self.x**2 + self.y**2) ** 0.5)

v1 = Vector(3, 4)
v2 = Vector(1, 2)

print(v1 + v2)   # → Vector(4, 6)
print(v1 - v2)   # → Vector(2, 2)
print(v1 * 3)    # → Vector(9, 12)
print(len(v1))   # → 5
```
