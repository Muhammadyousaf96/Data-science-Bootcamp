Encapsulation in OOP (Python)
===

What is Encapsulation?
Encapsulation means hiding the internal details of an object and only showing what is necessary. Think of it like a medicine capsule — the medicine is hidden inside, you just take it without knowing what's inside.
In Python, encapsulation means:

Wrapping data (variables) and methods (functions) together inside a class
Restricting direct access to some data


Access Modifiers in Python:

Public = name  ( Anywhere) 

Protected  =_name (Class + Subclass)

Private     =__name  (Class only)

Public
===

```python
class student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
    def show(self):
        print(f"the name:{self.name} age={self.age}")  
    
s=student("yousaf",21)
s.show()          
print(s.name)
    
```

private
===

```python
class BankAccount:
  def __init__(self, balance):
    self._balance = balance  # 🔐 Protected (single _)

  def show_balance(self):
    print(f"Balance: {self._balance}")

class SavingsAccount(BankAccount):  # child class
  def show(self):
    print(f"Savings: {self._balance}")  # ✅ OK in subclass

acc = BankAccount(5000)
acc.show_balance()             # ✅ Works
print(acc._balance)           # ⚠️ Works but NOT recommended

# Output:
Balance: 5000
5000
```

private
===

```python
class User:
  def __init__(self, name, password):
    self.name = name
    self.__password = password  # 🔒 Private (double __)

  def check_password(self, pwd):
    if self.__password == pwd:
      print("✅ Access Granted!")
    else:
      print("❌ Wrong Password!")

u = User("Ali", "secret123")
u.check_password("secret123")  # ✅ Works via method
# print(u.__password)           # ❌ ERROR! Cannot access

```

Getter & Setter 
===

• Getter = method to read a private variable safely

• Setter = method to update a private variable with validation

• This is the correct way to use encapsulation

• Prevents invalid data from entering (e.g., negative age, empty name)


```python
class student:
    def __init__(self,age):
        self.__age=age
    
    def get_age(self):
        return self.__age
    
    def set_age(self,age):
        if age>0:
            print(f"Age:{self.__age}")
        else:
            print("invalid age")
s=student(3)
s.get_age()  
s.set_age(-3)              
```
