## Algerian Forest Fires Dataset 
Data Set Information:

The dataset includes 244 instances that regroup a data of two regions of Algeria,namely the Bejaia region located in the northeast of Algeria and the Sidi Bel-abbes region located in the northwest of Algeria.

122 instances for each region.

The period from June 2012 to September 2012.
The dataset includes 11 attribues and 1 output attribue (class)
The 244 instances have been classified into fire(138 classes) and not fire (106 classes) classes.

imprt library

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
dataset=pd.read_csv("Algerian_forest_fires_dataset_UPDATE.csv",header=1)
```

### Data Cleaning

```python
# missing null values 
dataset[dataset.isnull().any(axis=1)]
```

```python
dataset.isnull().sum()
```

```python
dataset.loc[:122,"Region"]=0
dataset.loc[122:,"Region"]=1
df=dataset
```

```python
df.info()
```

```python
df[['Region']]=df[['Region']].astype(int)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
# removing the null value 
df=df.dropna().reset_index(drop=True)
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
# remove the 122 row
df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

#### fix space in columns 

```python
df.columns=df.columns.str.strip()
df.columns
```

```python
df.info()
```

### changes the required columns as integer data types 

```python
df.columns
```

```python
df.head(2)
# here we only change the { temperature ,RH,Ws} to interger from class
```

```python
df[['day', 'month', 'year','Temperature', 'RH', 'Ws']]=df[['day', 'month', 'year','Temperature', 'RH', 'Ws']].astype(int)
```

```python
df.info()
```

#### Here we change the {Rain,FFMC,DMC,DC,ISI,BUI,FWI	} into float and the class remain unchange 
#### Changing the other columns to float data datatype


```python
objects=[features for features in df.columns if df[features].dtype=='O']  #dtype='O' means object 
```

```python
for i in objects:
    if i!='Classes':
        df[i]=df[i].astype(float)
```

###  Monthly Fire Analysis

```python
## Monthly Fire Analysis
dftemp=df.loc[df['Region']==1]
plt.subplots(figsize=(10,5))
sns.set_style('whitegrid')
sns.countplot(x='month',hue='Classes',data=df)
plt.ylabel('Number of Fires',weight='bold')
plt.xlabel('Months',weight='bold')
plt.title("Fire Analysis of Sidi- Bel Regions",weight='bold')
plt.show()
```

```python
## Monthly Fire Analysis
dftemp=df.loc[df['Region']==0]
plt.subplots(figsize=(9,5))
sns.set_style('whitegrid')
sns.countplot(x='month',hue='Classes',data=df)
plt.ylabel('Number of Fires',weight='bold')
plt.xlabel('Months',weight='bold')
plt.title("Fire Analysis of Brjaia Regions",weight='bold')
plt.show()
```

Its observed that August and September had the most number of forest fires for both regions. And from the above plot of months, we can understand few things

Most of the fires happened in August and very high Fires happened in only 3 months - June, July and August.

Less Fires was on September
