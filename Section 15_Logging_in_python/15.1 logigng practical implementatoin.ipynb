python logging
===
Python's built-in logging module provides a flexible framework for emitting log messages from Python programs. It allows developers to track events that happen when some software runs, which can be crucial for debugging and monitoring applications.

Basic logging
===
To use the logging module, you can import it and configure it to your needs. Here's a simple example of how to log messages at different severity levels:

```python

````python
import logging 

# Basic configure 
logging.basicConfig(level=logging.DEBUG)

logging.debug("this is the debug message")
logging.error("this is the Error message")
logging.critical("this is the critical message")
logging.warning("Warning message")
logging.info("the info message")
````

lodgging to a file
===
You can also configure the logging module to write log messages to a file instead of the console. Here's how you can do that:

```python

````python
import logging 

logging.basicConfig(
    filename='app.log',
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d'
)
logging.info("application started")
logging.error("the database connectin failed")
````

logging with logger
===
For more complex applications, it's often better to use a logger object instead of the root logger. This allows you to have multiple loggers with different configurations. Here's an example:

```python


````python
import logging 

# crerate a named logger
logger=logging.getLogger(__name__)
logger.setLevel(logging.DEBUG)

# craete a handler
handler=logging.StreamHandler()
handler.setLevel(logging.DEBUG)

# creater a formatter 
formatter=logging.Formatter('%(asctime)s-%(name)s-%(levelname)s-%(message)s')
handler.setFormatter(formatter)

# add handler to logger
logger.addHandler(handler)

# use it 
logging.info("the user are logging")
logging.error('Database connection failed')
````
