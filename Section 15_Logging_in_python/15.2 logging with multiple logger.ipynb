logging with multiple loggers
====
In Python, you can use the `logging` module to create multiple loggers for different parts of your application. This allows you to have more control over the logging output and can help you organize your logs better.

```python
import logging

# create logger for module 1 
logger1=logging.getLogger("Module 1")
logger1.setLevel(logging.DEBUG)

# crate logger for module2
logger2=logging.getLogger("Module 2")
logger2.setLevel(logging.WARNING)

# configure logging setting
logging.basicConfig(
    level=logging.DEBUG,
    datefmt='%(asctime)s-%(name)s-%(message)s %H:%M:%S',
)
```

```python
# logging with different loggers
logger1.debug("this is module 1")
logger2.warning("module2 Waring ")

logger2.error("ereror in module2")
```
