A generator in Python is a special type of iterator that generates values one at a time using yield, instead of storing everything in memory.

🔁 What is a Generator?
===
Works like a function
Uses yield instead of return
Produces values lazily (on dema

```python
def square(n):
    for i in range(3):
      return i **2
```

```python
print(square(3))
```

```python
def square(n):
    for i in range(3):
      yield i **2
```

```python
sqr=square(3)
print(sqr)
```

```python
next(sqr)
```

```python
for i in square(3):
    print(i)
```

```python
def my_generater():
    yield 1
    yield 2
    yield 3
```

```python
gen=my_generater()
print(gen)
```

```python
for i in gen:
    print(i)
```

practical example : reading large files
===

```python
def read_large_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line
```

```python
file_path='large_file.txt'

for line in read_large_file(file_path):
    print(line.strip())
```
