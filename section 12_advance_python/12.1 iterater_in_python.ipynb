iterator in python 
===
An iterator is an object that can be iterated (looped) upon. It implements the iterator protocol,

 which consists of the methods __iter__() and __next__().

```python
list=[1,2,3,4,5,6,7,8]

iterater=iter(list)
```

```python
try:
   print(next(iterater)) 
except StopIteration:
    print("the list are empty and no more itertable")    
    
```
