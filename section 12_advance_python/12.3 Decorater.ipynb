Decorators
===
Decorators are flexible way to modify or extend behavior of functions or methods, without changing their actual code.

Decorators
===

```python
def main_welcom(func):
    
    def sub_welcom_method():
        print("this is the course of python")
        func()
        print("this is the course of advance python")
    return sub_welcom_method()
```

```python
def course_intro():
    print("this is new cours introductio ")
    
course_intro()
```

```python
main_welcom(course_intro)
```

```python
@main_welcom
def course_intro():
    print("this is new cours introductio 123 ")
```

Decorter with arguments
===

```python
def repeat(n):
    def decorator(func):
        def wrapper(*args,**kwargs):
            for i in range(n):
                func(*args,**kwargs)
        return wrapper        
    return decorator
```

```python
@repeat(3)
def sayHi():
    print("hiiii")
```

```python
sayHi()
```
