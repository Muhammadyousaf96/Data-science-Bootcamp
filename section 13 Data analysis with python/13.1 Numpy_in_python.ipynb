The Foundation of Scientific Python
===
NumPy (Numerical Python) is the fundamental package for numerical computing in Python. At its core, it introduces the ndarray — a powerful N-dimensional array object that is dramatically faster and more memory-efficient than Python's built-in lists.

```python
# Why NumPy for Data Analysis?
# Speed: NumPy operations run 10–100× faster than pure Python loops on numerical data.
# Memory efficiency: A NumPy array of 1 million integers uses ~8 MB; a Python list uses ~35 MB.
# Expressive syntax: Operations like "multiply every price by 1.1" are one-liners.
# Foundation of the ecosystem: Pandas, Matplotlib, Scikit-learn, and TensorFlow all build on NumPy arrays.
# Rich statistical functions: mean, median, std, percentile — all built in.
```

Python — why NumPy is faster
===

```python
import numpy as np
import time

# create one million number
data_list=list(range(1_000_000))
data_np=np.arange(1_000_000)

# python list loop aproach 
t1=time.time()
result=[x**2 for x in data_list]
print(f"pyhton list {time.time() -t1:.4f}s")

# numpay vectorze approach 
t2 = time.time()
result = data_np * 2
print(f"NumPy array: {time.time() - t2:.4f}s")   
```


Indexing & Slicing
===

Getting Exactly the Data You Need

In real analysis, you rarely need the whole dataset at once. Indexing lets you grab specific rows, columns, or elements. NumPy's slicing syntax is [start:stop:step] — same as Python lists, but extended to multiple dimensions.

2D Indexing: scores[row, column]


```python
score=np.array([[1,3,4,5],[45,67,88,5],[7,23,76,89]])
print(score)

print('------')

# slicing and indexing
print(score[1:])
print('-----')

print(score[2:])
```

Python — indexing & slicing


```python
scores = np.array([
    [85, 92, 78, 96],
    [70, 88, 91, 65],
    [55, 72, 68, 80],
])

# basic indexing 
print(scores[0,2])
print(scores[1])
print(scores[0:,1])
```

# ── Slicing ───
slicing is a way to grab a range of values from an array. It uses the syntax [start:stop:step], where start is the index to begin at, stop is the index to end before, and step is how many indices to skip. For example, arr[1:5] grabs indices 1 through 4, while arr[::2] grabs every other element.
---87----

indexing is a way to grab specific values from an array. It uses the syntax [index1, index2, ...], where each index corresponds to a dimension of the array. For example, arr[0] grabs the first element of a 1D array, while arr[0, 1] grabs the element in the first row and second column of a 2D array.

```python
#  Slicing
print(scores)

scores[0:2,1:3]  # rows 0-1, cols 1-2 → [[92,78],[88,91]]

scores[:,1:3] # all rows, cols 1 and 2

scores[::2]   # every other row
print('------------')
scores[:, ::-1]   # reverse column order

```

```python
# ── Boolean (Fancy) Indexing — very powerful! ──────

marks = np.array([45, 82, 59, 90, 71, 38, 95])

# Who passed? (marks >= 60)
passed=marks[marks>60]
print(passed)

# Who scored between 60 and 85?
scored=marks[(marks>=60) & (marks<=85)]
print(scored)

# Replace failing marks with 0 (data cleaning!)

marks[marks<60]=0
print(marks)

# ── Fancy indexing with arrays ─

prices = np.array([200, 450, 125, 780, 310])
selected = prices[[0, 2, 4]] 
print(selected)
```


Data Types (dtype)
===
Why Data Types Matter in Analysis

Every NumPy array has a single data type (dtype). Unlike Python where a list can hold mixed types, NumPy arrays are homogeneous — all elements share one type. This is what makes them fast, but you need to choose types wisely.



```python
# Python — working with dtypes

a=np.array([1,4,56,7])
print(a.dtype)
b=np.array([1,2,3,4], dtype=np.float64)
print(b.dtype)
```

Python — statistical operations (the heart of data analysis)
===

```python
marks = np.array([[85, 92, 78, 96],
                    [70, 88, 91, 65],
                    [55, 72, 68, 80]])
# ── Global statistics ──────────────────────────────
print(marks)
print(np.mean(marks))       # 78.33 — overall class average
print(np.median(marks))     # 80.0
print(np.std(marks))        # standard deviation
print(np.min(marks))        # 55 — lowest score overall
print(np.max(marks))        # 96 — highest score
print(np.percentile(marks, 75))  # 75th percentile = 90.25
```

```python
# ── Axis-based statistics (super useful!) ──────────
# axis=0 → collapse rows (result per COLUMN = per subject)
subject_avg = np.mean(marks, axis=0)
print(subject_avg)   # [70.0, 84.0, 79.0, 80.3] — avg per subject

# axis=1 → collapse columns (result per ROW = per student)
student_avg = np.mean(marks, axis=1)
print(student_avg)   # [87.75, 78.5, 68.75] — avg per student

# ── Finding positions ──────────────────────────────
print(np.argmax(student_avg))  # 0 — student 0 has highest avg
print(np.argsort(student_avg)) # [2, 1, 0] — sorted ranking

# ── Cumulative operations ──────────────────────────
monthly_sales = np.array([100, 150, 120, 200])
print(np.cumsum(monthly_sales)) # [100, 250, 370, 570] — running total
```

Broadcasting 
====
Operations Between Arrays of Different Shapes

Broadcasting is NumPy's mechanism for performing operations on arrays with different shapes — without copying data. It's one of NumPy's most powerful (and initially confusing) features.

The rule: Arrays are compatible if, starting from the trailing dimensions, their sizes either match or one of them is 1.

```python
# # Example: Subtract the column mean from each row
# marks shape: (3, 4) ← 3 students, 4 subjects
# col_mean shape: (1, 4) → (3, 4) ← broadcast across rows
```

![alt text](image.png)

```python
scores = np.array([[85, 92, 78, 96],
                    [70, 88, 91, 65],
                    [55, 72, 68, 80]])  # shape (3, 4)

# ── Scalar broadcasting (simplest) ────────────────
scores_scaled = scores * 1.1   # scalar → broadcast to (3,4)

# ── Row vector: normalize each subject ────────────
col_mean = np.mean(scores, axis=0)    # shape (4,)
centered = scores - col_mean           # (3,4) - (4,) → works!

# ── Column vector: subtract each student's avg ────
row_mean = np.mean(scores, axis=1, keepdims=True)  # shape (3,1) ← KEY!
deviation = scores - row_mean              # (3,4) - (3,1) → works!
print(deviation)  # how far each score is from that student's mean

# ── Practical: apply different bonuses per subject ─
subject_weights = np.array([0.3, 0.3, 0.2, 0.2])  # (4,)
weighted_scores = scores * subject_weights          # (3,4) × (4,)
final_grade     = weighted_scores.sum(axis=1)        # (3,)
print(final_grade)  # [87.4, 79.7, 68.7]
```

Mini project 
===

Student Examination Analysis System

You've been given marks data for 8 students across 5 subjects. Your task is to build a complete analysis pipeline using only NumPy — calculate stats, assign grades, identify top performers, and normalize scores.

```python
import numpy as np

# ════════════════════════════════════════════════
# DATASET: 8 students × 5 subjects
# Subjects: Math, Physics, Chemistry, CS, English
# ════════════════════════════════════════════════
marks = np.array([
    [88, 76, 91, 95, 82],   # Ali
    [72, 85, 68, 90, 77],   # Sara
    [55, 60, 48, 70, 65],   # Hamza
    [95, 92, 88, 98, 91],   # Ayesha
    [40, 38, 55, 45, 52],   # Usman
    [78, 82, 75, 80, 88],   # Zara
    [65, 70, 72, 60, 68],   # Bilal
    [82, 88, 85, 79, 90],   # Fatima
], dtype=np.float64)

students = ["Ali", "Sara", "Hamza", "Ayesha",
            "Usman", "Zara", "Bilal", "Fatima"]
subjects = ["Math", "Physics", "Chem", "CS", "Eng"]

# ── STEP 1: Basic Statistics ──────────────────────
student_avg  = np.mean(marks, axis=1)     # per student
student_max  = np.max(marks, axis=1)
student_min  = np.min(marks, axis=1)
student_std  = np.std(marks, axis=1)
subject_avg  = np.mean(marks, axis=0)     # per subject

print("=== STUDENT AVERAGES ===")
for i, name in enumerate(students):
    print(f"  {name:8s}: {student_avg[i]:.1f}  (std: {student_std[i]:.1f})")

print("\n=== SUBJECT AVERAGES ===")
for i, subj in enumerate(subjects):
    print(f"  {subj:10s}: {subject_avg[i]:.1f}")

# ── STEP 2: Grade Assignment using np.where ───────
def assign_grade(avg):
    return np.where(avg >= 90, 'A',
           np.where(avg >= 80, 'B',
           np.where(avg >= 70, 'C',
           np.where(avg >= 60, 'D', 'F'))))

grades = assign_grade(student_avg)
print("\n=== GRADES ===")
for i, name in enumerate(students):
    print(f"  {name:8s}: {grades[i]}  ({student_avg[i]:.1f})")

# ── STEP 3: Rankings ──────────────────────────────
rank_order = np.argsort(student_avg)[::-1]  # descending
print("\n=== STUDENT RANKINGS ===")
for rank, idx in enumerate(rank_order, 1):
    print(f"  #{rank}. {students[idx]:8s}: {student_avg[idx]:.1f}")

# ── STEP 4: Identify At-Risk Students ─────────────
at_risk = student_avg < 60
print("\n=== AT-RISK STUDENTS ===")
at_risk_names = [students[i] for i in np.where(at_risk)[0]]
print(f"  {at_risk_names}")

# ── STEP 5: Score Normalization (0-100 scale) ─────
# Min-max normalize each subject independently
col_min = marks.min(axis=0)
col_max = marks.max(axis=0)
normalized = (marks - col_min) / (col_max - col_min) * 100
print("\n=== NORMALIZED SCORES (first 3 students) ===")
print(np.round(normalized[:3], 1))

# ── STEP 6: Class-wide summary ────────────────────
print("\n=== CLASS SUMMARY ===")
print(f"  Overall average: {np.mean(marks):.1f}")
print(f"  Pass rate:       {(student_avg >= 60).mean()*100:.0f}%")
print(f"  Top scorer:      {students[np.argmax(student_avg)]}")
print(f"  Hardest subject: {subjects[np.argmin(subject_avg)]}")
print(f"  Easiest subject: {subjects[np.argmax(subject_avg)]}")

# ── STEP 7: Correlation between subjects ──────────
corr_matrix = np.corrcoef(marks.T)  # .T gives subject-subject corr
print(f"\n  Math-CS correlation: {corr_matrix[0,3]:.3f}")  # expect high
```
