The Foundation of Scientific Python
===
NumPy (Numerical Python) is the fundamental package for numerical computing in Python. At its core, it introduces the ndarray — a powerful N-dimensional array object that is dramatically faster and more memory-efficient than Python's built-in lists.

```python
# Why NumPy for Data Analysis?
# Speed: NumPy operations run 10–100× faster than pure Python loops on numerical data.
# Memory efficiency: A NumPy array of 1 million integers uses ~8 MB; a Python list uses ~35 MB.
# Expressive syntax: Operations like "multiply every price by 1.1" are one-liners.
# Foundation of the ecosystem: Pandas, Matplotlib, Scikit-learn, and TensorFlow all build on NumPy arrays.
# Rich statistical functions: mean, median, std, percentile — all built in.
```

Python — why NumPy is faster
===

```python
import numpy as np
import time

# create one million number
data_list=list(range(1_000_000))
data_np=np.arange(1_000_000)

# python list loop aproach 
t1=time.time()
result=[x**2 for x in data_list]
print(f"pyhton list {time.time() -t1:.4f}s")

# numpay vectorze approach 
t2 = time.time()
result = data_np * 2
print(f"NumPy array: {time.time() - t2:.4f}s")   
```
