What is Matplotlib?
===
Matplotlib is Python's most popular library for creating charts and graphs. You import it as import matplotlib.pyplot as plt and use it to turn raw data (like your student CSV) into visual insights.

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6]
y=[10,12,13,15,17,20]

plt.plot(x,y)
plt.title("Basic plot chart")
plt.xlabel('X axis')
plt.ylabel("y axis")
```

```python
# c'reate automated plot chart
x=[1,2,3,4,5,6]
y=[10,12,13,15,17,20]
plt.title('Automated plot chart')
plt.xlabel("x axis ")
plt.ylabel("y axis")
plt.plot(x,y, color='red',linestyle='--',marker='*',linewidth=4,markersize=20)
plt.grid()
```

Multiple plots
===
---Matplotlib allows you to create multiple plots in a single figure using the `subplot` function. This is useful for comparing different datasets or visualizing different aspects of the same dataset side by side.

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y1=[2,4,5,7,8]
y2=[10,13,15,17,19]

plt.Figure(figsize=(9,5))

plt.subplot(2,2,1)
plt.plot(x,y1, color='red')
plt.title("multi plot")

plt.subplot(1,2,2)
plt.plot(x,y2, color='blue')
plt.table("plot 2")

plt.subplot(2,2,3)
plt.plot(x,y2, color='blue')
plt.table("plot 3")

plt.subplot(2,2,2)
plt.plot(x,y2, color='blue')
plt.table("plot 4")
```

scatter plots
=== 
---Scatter plots are used to visualize the relationship between two continuous variables. You can create a scatter plot in Matplotlib using the `scatter` function, which takes the x and y coordinates of the points as arguments.

```python
import pandas as pd

df=pd.DataFrame({
    'Math':[30,40,50,60,70,44,45,66,76],
    'CS':[60,70,88,89,90,56,78,90,99]
})


math = df['Math']
cs   = df['CS']

plt.figure(figsize=(7, 5))
plt.scatter(math, cs,
            c='steelblue',   # color
            s=60,             # marker size
            alpha=0.7,        # transparency
            edgecolors='white',
            linewidths=0.5)
plt.xlabel('Math score')
plt.ylabel('CS score')
plt.title('Math vs CS correlation')
plt.show()
```

Histograms
===
---Histograms are used to visualize the distribution of a single continuous variable. You can create a histogram in Matplotlib using the `hist` function, which takes the data and the number of bins as arguments.

```python
import matplotlib.pyplot as plt
data=[4,5,6,6,7,8,8,9,9,9]

plt.hist(data,bins=5,color='gray',edgecolor='black')
# plt.xlabel("x asis")
```

pie charts
===
---Pie charts are used to visualize the proportions of different categories in a dataset. You can create a pie chart in Matplotlib using the `pie` function, which takes the sizes of the categories and their labels as arguments.

```python
grade=['A','B','C','D']
count=[10,7,5,1]

explode=(0.1,0,0,0.06)
colors=['gold','silver','green','red']
```

```python
# create a pie chart
plt.pie(count,labels=grade,colors=colors,explode=explode,autopct="%1.1f%%",shadow=5)
```
