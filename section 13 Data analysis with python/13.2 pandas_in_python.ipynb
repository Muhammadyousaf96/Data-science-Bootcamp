pandas for data analysis in Python:
===
pandas is a powerful library for data manipulation and analysis. It provides two main data structures: Series (1D) and DataFrame (2D). With pandas, you can easily load, clean, explore, and visualize data. It offers a wide range of functions for filtering, grouping, and summarizing data, making it an essential tool for any data analyst or scientist working in Python.

serires:
===
A Series is a one-dimensional labeled array that can hold any data type. It consists of two main components: the index and the values. The index is a set of labels that uniquely identifies each element in the Series, while the values are the actual data stored in the Series. You can create a Series from a list, dictionary, or even a NumPy array. Series are useful for handling time series data, categorical data, or any situation where you want to associate labels with your data.

```python
# series 
import pandas as pd 
import numpy as np

# ── Create a Series ──────────────────────────────
marks=pd.Series([2,3,4,5,6])
# print(marks)
# print(type(marks))

index=['a','c','f','s','r']
data=[3,4,5,6,7]
pd.Series(data,index=index)

score=pd.Series([4,5,6,7],index=['d','r','t','o'])
print(score)

score['d']
score.iloc[3]

# print(score['t'])

# score[score > 4]

# print(score.mean())

# Useful Series properties
score.values # underlying NumPy array
score.index    # Index(['Ali','Sara','Hamza','Ayesha','Usman'])
score.dtype    # int64
score.name     # None (you can set: marks.name = 'Score')
```

Dataframe:
===
A DataFrame is a two-dimensional labeled data structure that can hold data of different types (e.g., integers, floats, strings). It is similar to a table in a relational database or an Excel spreadsheet. A DataFrame consists of rows and columns, where each column can be thought of as a Series. You can create a DataFrame from various sources such as dictionaries, lists, or even other DataFrames. DataFrames are ideal for handling structured data and performing operations like filtering, grouping, and merging datasets.

# Cleaning Messy Data
80% of real data science work is cleaning. Pandas is built for this.


Handling Missing Values (NaN)

Real data is almost always incomplete. NaN (Not a Number) is Pandas' representation of missing data. Handling it correctly is critical — wrong decisions here propagate errors through your entire analysis.

```python
import seaborn as sns
df = sns.load_dataset('titanic')

df.head(5)
```

# ── Detect ─────────

```python

# df.isnull()               # True/False mask for every cell
df.isnull().sum()         # total nulls per column
# df.notnull()              # opposite
df[df['embark_town'].isnull()]  # rows where revenue is missing
```

# ── Drop ───


```python
df.dropna()              # drop any row with at least 1 NaN
#df
df.dropna(axis=1)         # drop any COLUMN with at least 1 NaN

df.dropna(subset='age')    # drop rows where 'age' is NaN
df.dropna(thresh=4).head(2)      # keep rows with at least 4 non-null values
```

Fill
===

```python
# df.fillna(0)                          # fill ALL NaN with 0
df['age'].fillna(df['age'].mean())      #fill with column mean 

df['embark_town'].fillna('Unknown')     # fill categorical with a label
df['age'].fillna('Unknown')

df.fillna(method='ffill')                # forward-fill (use previous value)

df.fillna(method='bfill').head(3)                 # backward-fill

# df.interpolate()                       # linear interpolation  
```

```python
# ── Fill different columns differently ────────────

df.fillna({'age': df['age'].median(),
           'region':  'Unknown',
           'units':   0
           }).head(3)
```

Filtering & Sorting
===
Query your data like a database — get exactly the rows you need, in the right order.

```python
df=pd.read_csv('students.csv')
```

```python
# ── Boolean filtering ─────────────────────────────
df[df['Math'] > 80]                         # high math scores
df[(df['Math'] > 80) & (df['CS'] > 90)]   # AND: both conditions
df[(df['Math'] < 60) | (df['CS'] < 60)]   # OR: either condition
df[~(df['Grade'] == 'F')].head(2)                  # NOT: negate
```

```python
# # ── .query() — cleaner syntax ─────────────────────
df.query('Math > 80 and CS > 85')
df.query('Grade in ["A", "B"]').head(2)
# df.query('Math > @threshold')   # @ references a Python variable
```

```python
# ── .isin() — match a list of values ──────────────
df[df['region'].isin(['North', 'South'])]
df[~df['region'].isin(['Unknown'])].head(2)  # exclude 'Unknown'
```

```python
# ── .between() — range filter ─────────────────────
df[df['revenue'].between(100, 500)]   # inclusive
```

```python
# # ── Sorting ───────────────────────────────────────
df.sort_values('Math')                          # ascending
df.sort_values('Math', ascending=False)        # descending
df.sort_values(['Grade', 'Math'],)             # multi-key sort
#                ascending=[True, False])
df.sort_values('Math').head(5)                # bottom 5
df.nlargest(5, 'revenue')                     # top 5 by revenue (faster)
df.nsmallest(3, 'Math')                       # 3 lowest math scores
```
