sqlite3
===
SQLite is a C library that provides a lightweight disk-based database that doesn’t require a separate server process and allows access to the database using a nonstandard variant of the SQL query language. Some applications can use SQLite for internal data storage. It’s also possible to prototype an application using SQLite and then port the code to a larger database such as PostgreSQL or Oracle.

```python
import sqlite3 
```

```python
# connet to sqlite database
connection=sqlite3.connect('example.db')

cursor=connection.cursor()
```

```python
# create a table 
cursor.execute('''
Create Table If Not Exists employees(
    id Interger Primary Key,
    name Text not Null,
    age Integer,
    department Text
)               
''')

# commit the changes
connection.commit()
```

```python
# select 
cursor.execute('''
Select *from employees
''')
```

```python
# INsert  data into sqlite table 
cursor.execute(
'''
Insert Into employees(age,name,department)
Values('asad',31,'sc')               
''')

cursor.execute(
'''
Insert Into employees(age,name,department)
Values('Muhammad Farhan',22,'conputer science')               
''')

cursor.execute(
'''
Insert Into employees(age,name,department)
Values('Asad',23,'Chemistry')               
''')
connection.commit()
```

```python
# queery the data form table 
cursor.execute('Select *from employees')
rows=cursor.fetchall()

# print the queery data 
for row in rows:
    print(row)
```

```python
# Working with sale data 
import sqlite3
conn=sqlite3.connect('sales_data.db')
cursor=conn.cursor()

# crete table for sale data 
cursor.execute('''
CREATE TABLE IF NOT EXISTS sales(
    id INTEGER PRIMARY KEY,
    date TEXT NOT NULL,
    product TEXT NOT NULL,
    sales INTEGER,
    region TEXT
)
''')
conn.commit()

# insert data into sale table 

sales_data=[
    (2023-4-3,'product1',200,'North'),
    (2023-5-3,'product2',300,'WEST'),
    (2023-4-3,'product3',600,'South'),
    (2023-4-3,'product4',800,'Est'),
]
conn.commit()

# insert the sales data into the table 
cursor.executemany('''
Insert into sales(date,product,sales,region)
               values(?,?,?,?)
''',sales_data)

conn.commit()

# querry the data 
cursor.execute('Select *from sales')
rows=cursor.fetchall()
# print
for row in rows:
    print(row)
```

```python
# close the connection
cursor.close()
```
