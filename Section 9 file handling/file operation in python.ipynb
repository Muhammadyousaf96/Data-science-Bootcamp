#commom mode
| Mode   | Meaning                                |
| ------ | -------------------------------------- |
| `"r"`  | Read (default) – file must exist       |
| `"w"`  | Write – creates new file or overwrites |
| `"a"`  | Append – add data at end               |
| `"x"`  | Create a new file – error if exists    |
| `"r+"` | Read + Write                           |
| `"b"`  | Binary mode (`rb`, `wb`)               |
| `"t"`  | Text mode (default)                    |
